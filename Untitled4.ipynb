```python
import tkinter as tk
from tkinter import messagebox

CELL_SIZE = 60

class NQueensUI:
    def __init__(self, root):
        self.root = root
        self.root.title("N-Queens Solver")

        self.n = 0
        self.solutions = []
        self.current_solution = 0
        self.canvas = None

        self.build_controls()

    def build_controls(self):
        frame = tk.Frame(self.root)
        frame.pack(pady=10)

        tk.Label(frame, text="Enter N:").pack(side=tk.LEFT)

        self.n_entry = tk.Entry(frame, width=5)
        self.n_entry.pack(side=tk.LEFT, padx=5)

        tk.Button(frame, text="Solve", command=self.solve).pack(side=tk.LEFT, padx=5)
        tk.Button(frame, text="Next", command=self.show_next).pack(side=tk.LEFT)

        self.info_label = tk.Label(self.root, text="")
        self.info_label.pack(pady=5)

    def draw_board(self):
        if self.canvas:
            self.canvas.destroy()

        size = self.n * CELL_SIZE
        self.canvas = tk.Canvas(self.root, width=size, height=size)
        self.canvas.pack()

        for row in range(self.n):
            for col in range(self.n):
                color = "white" if (row + col) % 2 == 0 else "gray"
                x1 = col * CELL_SIZE
                y1 = row * CELL_SIZE
                x2 = x1 + CELL_SIZE
                y2 = y1 + CELL_SIZE
                self.canvas.create_rectangle(x1, y1, x2, y2, fill=color)

    def draw_queens(self, board):
        self.canvas.delete("queen")
        for row, col in enumerate(board):
            x = col * CELL_SIZE + CELL_SIZE // 2
            y = row * CELL_SIZE + CELL_SIZE // 2
            self.canvas.create_text(
                x, y, text="♛", font=("Arial", 32), fill="red", tags="queen"
            )

    def is_safe(self, board, row, col):
        for r in range(row):
            c = board[r]
            if c == col or abs(c - col) == abs(r - row):
                return False
        return True

    def solve_all(self, board, row):
        if row == self.n:
            self.solutions.append(board.copy())
            return

        for col in range(self.n):
            if self.is_safe(board, row, col):
                board[row] = col
                self.solve_all(board, row + 1)
                board[row] = -1

    def solve(self):
        try:
            self.n = int(self.n_entry.get())
            if self.n < 4:
                raise ValueError
        except ValueError:
            messagebox.showerror("Error", "Please enter an integer N ≥ 4")
            return

        self.solutions.clear()
        self.current_solution = 0

        board = [-1] * self.n
        self.solve_all(board, 0)

        if not self.solutions:
            messagebox.showinfo("Result", "No solutions found")
            return

        self.draw_board()
        self.draw_queens(self.solutions[0])

        self.info_label.config(
            text=f"Total Solutions: {len(self.solutions)} | Showing: 1"
        )

    def show_next(self):
        if not self.solutions:
            return

        self.current_solution = (self.current_solution + 1) % len(self.solutions)
        self.draw_queens(self.solutions[self.current_solution])

        self.info_label.config(
            text=f"Total Solutions: {len(self.solutions)} | Showing: {self.current_solution + 1}"
        )


if __name__ == "__main__":
    root = tk.Tk()
    app = NQueensUI(root)
    root.mainloop()
```

### N-Queens Solver (Console Version for Colab)

This version of the N-Queens solver runs in Google Colab by printing the solutions directly to the console instead of using a graphical user interface.

```python
def is_safe(board, row, col, n):
    # Check this row on left side
    for i in range(col):
        if board[row][i] == 1:
            return False

    # Check upper diagonal on left side
    for i, j in zip(range(row, -1, -1), range(col, -1, -1)):
        if board[i][j] == 1:
            return False

    # Check lower diagonal on left side
    for i, j in zip(range(row, n, 1), range(col, -1, -1)):
        if board[i][j] == 1:
            return False
    return True

def solve_n_queens_util(board, col, n, solutions):
    # Base case: If all queens are placed then return true
    if col >= n:
        solutions.append([row[:] for row in board]) # Deep copy the board
        return True

    # Consider this column and try placing this queen in all rows one by one
    res = False
    for i in range(n):
        if is_safe(board, i, col, n):
            board[i][col] = 1 # Place queen
            res = solve_n_queens_util(board, col + 1, n, solutions) or res
            board[i][col] = 0 # Backtrack (remove queen)
    return res

def solve_n_queens(n):
    board = [[0 for _ in range(n)] for _ in range(n)]
    solutions = []
    solve_n_queens_util(board, 0, n, solutions)
    return solutions

def print_board(board):
    for row in board:
        print(" ".join(["Q" if cell == 1 else "." for cell in row]))

# --- Main execution ---

# You can change the value of N here
N_VALUE = 4

if N_VALUE < 4:
    print("N must be an integer greater than or equal to 4 for non-trivial solutions.")
else:
    print(f"Solving N-Queens for N = {N_VALUE}")
    all_solutions = solve_n_queens(N_VALUE)

    if not all_solutions:
        print("No solutions found.")
    else:
        print(f"Found {len(all_solutions)} solutions.")
        for i, solution in enumerate(all_solutions):
            print(f"\nSolution {i + 1}:")
            print_board(solution)
```

## Implement Board Visualization

### Subtask:
Create a function that takes a solved N-Queens board (a list of queen positions) and generates a visual representation of it using `ipywidgets.HTML`.


**Reasoning**:
I need to implement the `display_board_html` function as described in the instructions. This involves iterating through the board, creating HTML for each cell with alternating colors and queen symbols, applying CSS styling, and finally wrapping it in an `ipywidgets.HTML` object.



```python
def display_board_html(solution_board, n):
    cell_size = 50  # pixels
    board_html_rows = []

    for r in range(n):
        row_cells = []
        for c in range(n):
            # Determine cell color for chessboard pattern
            bg_color = "#D18B47" if (r + c) % 2 == 0 else "#FFCE9E"

            # Place queen or empty cell
            queen_char = "&#x265B;" if solution_board[r][c] == 1 else ""

            # CSS for each cell
            cell_style = f"width: {cell_size}px; height: {cell_size}px; background-color: {bg_color}; " \
                         "display: inline-flex; justify-content: center; align-items: center; " \
                         "border: 1px solid #333; font-size: {cell_size * 0.7}px;"

            row_cells.append(f"<div style='{cell_style}'>{queen_char}</div>")
        board_html_rows.append("<div style='display: flex;'>" + "".join(row_cells) + "</div>")

    # Overall board styling
    board_style = f"border: 2px solid #333; display: inline-block; line-height: 0;"
    full_html = f"<div style='{board_style}'>{''.join(board_html_rows)}</div>"

    return widgets.HTML(full_html)

print("The 'display_board_html' function has been defined.")
```

## Build Interactive Controls

### Subtask:
Develop `ipywidgets` components for user interaction. This includes an input field for 'N', a 'Solve' button, and 'Next Solution' / 'Previous Solution' buttons to navigate through the found solutions. An `Output` widget will be used to display the board and solution count.


**Reasoning**:
I need to initialize the `ipywidgets` components as per the instructions: an `IntText` for N, three `Button` widgets, an `Output` widget, and an `HTML` widget for solution information. Then, I will print a message confirming their setup.



```python
n_input = widgets.IntText(
    value=4,
    description='Enter N:',
    disabled=False,
    min=4,
    max=12 # Set a reasonable max for demonstration
)

solve_button = widgets.Button(
    description='Solve',
    disabled=False,
    button_style=''
)

next_button = widgets.Button(
    description='Next',
    disabled=False,
    button_style=''
)

prev_button = widgets.Button(
    description='Previous',
    disabled=False,
    button_style=''
)

output_widget = widgets.Output()
solution_info_label = widgets.HTML(value="Total Solutions: 0 | Showing: 0")

print("Interactive controls (N input, Solve, Next, Previous buttons, Output widget, and Info label) have been initialized.")
```

## Connect Logic to GUI

### Subtask:
Write callback functions that link user actions (like clicking buttons or changing the value of N) to the N-Queens solving logic and the board visualization.


**Reasoning**:
I will initialize global variables, define callback functions for the 'Solve', 'Next', and 'Previous' buttons, and then link these functions to the button widgets to enable interactive control of the N-Queens solver.



```python
all_solutions_interactive = []
current_solution_index = 0

def solve_n_queens_interactive(b):
    global all_solutions_interactive, current_solution_index

    with output_widget:
        output_widget.clear_output()
        try:
            n = n_input.value
            if n < 4:
                raise ValueError("N must be an integer >= 4.")
        except ValueError as e:
            print(f"Error: {e}")
            solution_info_label.value = "Total Solutions: 0 | Showing: 0"
            return

        print(f"Solving N-Queens for N = {n}...")

        # Use the already defined core solver logic (from cell 6eaf3746)
        # Ensure these functions are available in the current scope if they were in a previous cell
        # For this execution, we assume they are already defined.
        board_init = [[0 for _ in range(n)] for _ in range(n)]
        temp_solutions = []
        solve_n_queens_util(board_init, 0, n, temp_solutions)

        all_solutions_interactive = temp_solutions
        current_solution_index = 0

        if not all_solutions_interactive:
            solution_info_label.value = "No solutions found."
            print("No solutions found.")
        else:
            solution_info_label.value = f"Total Solutions: {len(all_solutions_interactive)} | Showing: {current_solution_index + 1}"
            display(display_board_html(all_solutions_interactive[current_solution_index], n))

def show_next_solution_interactive(b):
    global current_solution_index

    if not all_solutions_interactive:
        return

    current_solution_index = (current_solution_index + 1) % len(all_solutions_interactive)

    with output_widget:
        output_widget.clear_output()
        n = n_input.value
        display(display_board_html(all_solutions_interactive[current_solution_index], n))
        solution_info_label.value = f"Total Solutions: {len(all_solutions_interactive)} | Showing: {current_solution_index + 1}"

def show_prev_solution_interactive(b):
    global current_solution_index

    if not all_solutions_interactive:
        return

    current_solution_index = (current_solution_index - 1 + len(all_solutions_interactive)) % len(all_solutions_interactive)

    with output_widget:
        output_widget.clear_output()
        n = n_input.value
        display(display_board_html(all_solutions_interactive[current_solution_index], n))
        solution_info_label.value = f"Total Solutions: {len(all_solutions_interactive)} | Showing: {current_solution_index + 1}"

# Attach callback functions to buttons
solve_button.on_click(solve_n_queens_interactive)
next_button.on_click(show_next_solution_interactive)
prev_button.on_click(show_prev_solution_interactive)

print("Callback functions defined and attached to buttons.")
```
